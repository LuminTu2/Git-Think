<p><font size="4"><strong>概述</strong></font>
<br><br>
<strong>K-Nearest Neighbors</strong>是一个有监督学习算法。该方法的主要思想是：如果一个样本在特征空间中的k个最相似(即特征空间中最邻近)的样本中的大多数属于某一个类别，则该样本也属于这个类别。即是给定一个训练数据集，对新的样本属于哪一类，在训练数据集中找到与该样本最邻近的K个样本，这K个样本的多数所属的类，即可认为新样本也属于这一类。</p>

<p><strong>下面是K-Nearest Neighbors算法的可视化。</strong></p>


<p>在这种情况下，我们有A类和B类数据点。我们想要预测星（测试数据点）属于哪一类。 如果我们考虑k值为3（3个最近的数据点），星将被预测属于B类。但是如果我们考虑k值为6，星将被预测属于A类。</p>
<p>从这个意义上讲，考虑k的值很重要。希望从这个图中，您可以了解K-最近邻居算法是什么。 </p>

<p><font size="4"><strong>算法</strong></font></p>
<p>KNN的工作原理是：存在一个样本数据集合，也称作训练样本集，并且样本集中每个数据都存在标签，即我们知道样本集中每一数据与所属分类的对应关系。输入没有标签的新数据后，将新数据的每个特征与样本集中数据对应的特征进行比较，然后算法提取样本集中特征最相似数据（最近邻）的分类标签。一般来说，我们只选择样本数据集中前k个最相似的数据，通常k是不大于20的整数，最后，选择k个最相似数据中出现次数最多的分类，作为新数据的分类。</p>

    <li> 数据集内有非数字数据 </li>
    <li> 行号会对我们建模造成影响 </li>
</ul>

<p>为了解决这些问题，需要定义函数<b> removeColumns(DataFrame, column) </b>：</p>
<blockquote>
<p>输入参数:</p>
<blockquote>
<p><code>Dataframe</code>：一个包含数据特征和标签的Dataframe(my_data)<br>
<code>column</code>：想要移除的列的编号，注意Python是从0开始索引，因此第一列的编号将为0</p>
</blockquote>
</blockquote>
<blockquote>
<p>输出：</p>
<blockquote>
<p><code>del_df</code>：返回一个没有指定列的numpy数组 </p>
</blockquote>
</blockquote>
</div>

<div class="alert alert-success alertsuccess" style="margin-top: 5px">
代码示例：
</div>

```python
# 删除包含目标名称的列，因为它不包含数值。
# 同时删除包含行号的列
# axis = 1表示我们正在删除列而不是行。
# 函数输入一个DataFrame和列值同时返回没有起始列的numpy数组。
def removeColumns(DataFrame, *column):
    # axis = 1表示我们正在删除列而不是行
    del_df = DataFrame.drop(DataFrame.columns[[column]], axis=1).values
    return del_df
```

<div data-v-c9883c9c="" class="render-html"><p>现在，我们得到拟合模型所需的X，即new_data</p>
<p>接下来，我们需要得到响应向量y。 既然不能使用.target和.target_names，我们需要定义一个函数<b> targetAndtargetNames(Dataframe, targetColumnIndex)： </b></p>
<blockquote>
<p>输入参数:</p>
<blockquote>
<p><code>Dataframe</code>：一个包含数据特征和标签的Dataframe<br>
<code>targetColumnIndex</code>：表示目标列（target column）的编号（索引），注意Python是从0开始索引，因此第一列将为0</p>
</blockquote>
</blockquote>
<blockquote>
<p>输出：</p>
<blockquote>
<p><code>np.asarray(target)</code>：响应向量<br><code>target_names</code>：标签名称</p>
</blockquote>
</blockquote>
</div>

<div class="alert alert-success alertsuccess" style="margin-top: 5px">
代码示例：
</div>

```python
def targetAndtargetNames(Dataframe, targetColumnIndex):
    target_dict = dict()
    target = list()
    target_names = list()
    count = -1
    for i in range(len(Dataframe.values)):
        if Dataframe.values[i][targetColumnIndex] not in target_dict:
            count += 1
            target_dict[Dataframe.values[i][targetColumnIndex]] = count
        target.append(target_dict[Dataframe.values[i][targetColumnIndex]]) 
    #由于字典没有排序，因此我们需要对它进行排序并将其输出到列表中，以便目标名称与目标匹配。
    for targetName in sorted(target_dict, key=target_dict.get):
        target_names.append(targetName)
    return np.asarray(target), target_names
```

```python
#使用targetAndtargetNames函数, 创建两个变量，分别命名为target和target_names,并且打印它们
target, target_names = targetAndtargetNames(my_data, 1) 
print(target)
print(target_names)
```
